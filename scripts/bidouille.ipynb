```python
file = "/Users/cc54/Documents/Lab/Analyses/TEleVIZion/data/EDTA/GCA_965234655.1_gc_windows_10000.tsv"
sim = {"OZ248582.1":"2RL", "OZ248583.1":"3RL", "OZ248584.1":"X"}

for line in open(file).readlines():
    if line.startswith("chr"):
        txt = line
    else:
        chr, start, end, name, itemRgb = line[:-1].split("\t")
        if chr in sim:
            new_line = "\t".join([sim[chr], start, end, name, itemRgb]) + "\n"
            txt += new_line

with open(file, "w") as out:
    out.write(txt)
```
